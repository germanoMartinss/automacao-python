```python
import os
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
from nltk import sent_tokenize, word_tokenize
from nltk.probability import FreqDist
from nltk.corpus import stopwords
from wordcloud import WordCloud, STOPWORDS
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
with open(os.path.join('data', 'texto.txt'), 'r', encoding='utf-8-sig') as file:
    text = file.read()
text
```

```python
texto_full = " ".join(text.splitlines())
texto_full
```

```python
sent_tokens = sent_tokenize(texto_full)
sent_tokens
```

```python
word_tokens = word_tokenize(texto_full)
word_tokens
```

```python
len(word_tokens)
```

```python
fdist = FreqDist(word_tokens)
fdist
```

```python
fdist.most_common(10)
```

```python
step_words = set(stopwords.words('portuguese'))
```

```python
filtered_words = [word for word in word_tokens if word.isalnum() and word not in step_words]
filtered_words
```

```python
len(filtered_words)
```

```python
fdist2 = FreqDist(filtered_words)
fdist2.most_common(10)
```

```python
fdist2.plot(10, cumulative=False)
```

```python
def plot_cloud(wordcloud):
    plt.figure(figsize=(40,30))
    plt.imshow(wordcloud)
    plt.axis("off")
    plt.show()
    
```

```python
wc = WordCloud(
    width=3000,
    height=2000,
    random_state=1,
    background_color="white",
    colormap="Pastel1",
    collocations=False,
    stopwords=STOPWORDS
)
wc.generate(texto_full)
```

```python
plot_cloud(wc)
```

```python
imagem = Image.open(os.path.join('data', 'upvote.png')).convert('RGBA')
fundo = Image.new('RGBA', imagem.size, (255, 255, 255, 255))
imagem_composta = Image.alpha_composite(fundo, imagem).convert('RGB')
mascara = np.array(imagem_composta)
mascara
```

```python
wc2 = WordCloud(
    width=3000,
    height=2000,
    random_state=1,
    background_color="white",
    colormap="Pastel1",
    collocations=False,
    stopwords=STOPWORDS,
    mask=mascara
)
wc2.generate(texto_full)
```

```python
plot_cloud(wc2)
```
