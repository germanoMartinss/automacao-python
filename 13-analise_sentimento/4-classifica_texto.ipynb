```python
from transformers import pipeline
```

```python
classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")
```

```python
categories = ["Outdoor Gear", "Kitchen Appliances", "Apparel"]
```

```python
frases = [
    "I bought a lightweight, waterproof tent for this weekend's hiking trip.",  # Outdoor Gear
    "The new air fryer cooks fast and makes cleaning my kitchen so much easier.",  # Kitchen Appliances
    "This leather jacket goes with everything and the fabric is really comfortable.",  # Apparel
]
```

```python
for frase in frases:
    result = classifier(frase, categories)
    print(f"Frase: {frase}")
    print(f"Categoria: {result['labels'][0]}")
    print(f"Pontuação: {result['scores'][0]:.4f}")
    print("-" * 50)
```
