```python
from transformers import pipeline
```

```python
model_qa = pipeline("question-answering", model="deepset/roberta-base-squad2")
```

```python
context = """
Sarah decided to spend her Saturday afternoon at the local farmers
market, where she enjoyed sampling fresh produce and handmade crafts
"""

question = "Where did Sarah spend her Saturday afternoon?"
```

```python
context = " ".join(context.split())
result = model_qa(question=question, context=context)
print(result["answer"], result["score"])
```

```python
model_qa(question=question, context=context)
```

```python
import pandas as pd
```

```python
df_questions = pd.read_csv("dados/games_faq.csv")
df_questions
```

```python
def question_answering(question):
    context = df_questions[df_questions["question"] == question]
    result = model_qa(question=question, context=context["answer"].values[0])
    return result["answer"]
    # return context["answer"].values[0]
```

```python
question_answering("How do I unlock new levels in the game?")
```

```python
question_answering("Can I play the game offline?")
```
