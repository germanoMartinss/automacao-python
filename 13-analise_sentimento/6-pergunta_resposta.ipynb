```python
from transformers import pipeline
```

```python
model_qa = pipeline("question-answering", model="deepset/roberta-base-squad2")
```

```python
context = """
Sarah decided to spend her Saturday afternoon at the local farmers
market, where she enjoyed sampling fresh produce and handmade crafts
"""

question = "Where did Sarah spend her Saturday afternoon?"
```

```python
context = " ".join(context.split())
result = model_qa(question=question, context=context)
print(result["answer"], result["score"])
```
