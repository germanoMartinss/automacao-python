```python
from liqfit.pipeline import ZeroShotClassificationPipeline
from liqfit.models import T5ForZeroShotClassification
from transformers import T5Tokenizer
```

```python
model = T5ForZeroShotClassification.from_pretrained('knowledgator/comprehend_it-multilingual-t5-base')
tokenizer = T5Tokenizer.from_pretrained('knowledgator/comprehend_it-multilingual-t5-base')
classifier = ZeroShotClassificationPipeline(model=model, tokenizer=tokenizer,
                                                      hypothesis_template = '{}', encoder_decoder = True)
```

```python
categorias = ["Esporte", "Política", "Entretenimento"]
```

```python
frases = [
    "O time venceu o campeonato nos pênaltis e a torcida comemorou até de madrugada.",  # Esporte
    "O Senado aprovou o projeto de lei depois de uma longa votação e muita discussão.",  # Política
    "A nova série estreou ontem na plataforma de streaming e já é a mais assistida.",    # Entretenimento
]
```

```python
for frase in frases:
    result = classifier(frase, categorias)
    print(f"Frase: {frase}")
    print(f"Categoria: {result['labels'][0]}")
    print(f"Pontuação: {result['scores'][0]:.4f}")
    print("-" * 50)
```

```python
import pandas as pd
from transformers import pipeline
```

```python
df_frases = pd.read_csv("dados/frases_categorias.csv")
categorias = ["ESPORTE", "CIÊNCIA", "MODA"]
```

```python
resultados_df = classifier(df_frases['Frase'], categorias, multi_label=False)
```

```python
def classificar(frase):
    resultados_df = classifier(frase, categorias, multi_label=False)
    return resultados_df['labels'][resultados_df['scores'].index(max(resultados_df['scores']))]
```

```python
df_frases["Categoria"] = df_frases["Frase"].apply(classificar)
df_frases
```
